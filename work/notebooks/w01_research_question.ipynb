### 1. My Lane Selection
**Chosen Lane:** Lane 1 — Content Refresh Prioritization (Predicting Search Traffic Decline)

**Why this lane:**
In SEO and content marketing, deciding which existing pages to update is one of the highest-leverage actions a team can take. Updating dying content is significantly cheaper than producing new content from scratch. However, rule-based heuristics (like flagging every article older than 365 days) suffer from low precision and waste valuable writer hours on pages that are still performing well. Choosing this lane allows us to build a predictive framework that ranks declining URLs by intervention priority.

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### 2. Problem Framing & Decision Matrix
* **Unit of Analysis:** A single URL / web page on a client domain evaluated over a 90-day window.
* **Target Output:** Probability score $P(\text{declining})$ indicating the likelihood that organic traffic and rankings for the page will suffer a sustained drop.
* **The Decision:** Whether to queue a specific URL for an immediate editorial refresh / content update.
* **The Action:** An SEO strategist generates a content refresh brief for high-risk URLs, directing writers to update outdated facts, improve intent coverage, or expand target keywords.
* **Cost of a Wrong Call:**
  * **False Positive (Predicting decline when page is healthy):** Wastes $200–$500 in writer fees per article and risks disrupting a page that search engine algorithms are currently rewarding.
  * **False Negative (Predicting stability when page is dying):** Compounding loss of organic search impressions, ranking positions, and conversion revenue to competitors. Recovery becomes harder the longer the decay persists.
* **Why ML/Data helps:** Simple heuristics fail because traffic drop is multi-faceted. Machine learning synthesizes complex interactions between page age, impressions, click-through rates (CTR), and average search position to output a calibrated risk score rather than a blunt rule.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os
import pandas as pd

# Fallback direct URL for Google Colab environments
raw_github_url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"

# Relative paths for local repo runs
possible_paths = [
    "../../data/raw/content_refresh_anonymized.csv",
    "../data/raw/content_refresh_anonymized.csv",
    "data/raw/content_refresh_anonymized.csv"
]

# Check local paths first; use public URL if running standalone in Colab
data_source = next((p for p in possible_paths if os.path.exists(p)), raw_github_url)

df = pd.read_csv(data_source)

total_pages = len(df)
declining_pages = (df["trend_direction"] == "down").sum()
pct_declining = (declining_pages / total_pages) * 100

median_age_down = df[df["trend_direction"] == "down"]["content_age_days"].median()
median_age_up = df[df["trend_direction"] == "up"]["content_age_days"].median()

print("=== DATASET EVIDENCE ===")
print(f"1. Total URLs Analyzed: {total_pages:,}")
print(f"2. Baseline Declining Pages ('down' trend): {declining_pages:,} ({pct_declining:.1f}%)")
print(f"3. Median Content Age: Declining pages = {median_age_down:.0f} days vs. Growing pages = {median_age_up:.0f} days")
```

## 4. Careful words: what I can and can't claim

### 4. What I Can and Cannot Claim

**What I CAN claim:**
* Historical feature distributions (age, CTR, position trends) correlate with a page's likelihood of entering a decline phase.
* Predictive modeling can rank URLs by risk significantly better than random guessing or single-variable rules.

**What I CANNOT claim:**
* **No Causal Guarantee:** A high risk prediction does not prove that age or low word count caused the drop; Google core algorithm updates, competitor actions, or seasonality may be unobserved factors.
* **No Automatic Recovery:** Re-optimizing a flagged page does not guarantee a 100% traffic recovery—it merely increases the probability of arresting traffic loss compared to doing nothing.
